iloc and loc

The core difference between .loc and .iloc comes down to labels vs. positions."

".iloc is integer-position based. It selects data based on its 0-indexed numerical location, exactly like standard Python list slicing. For example, df.iloc[0:3] will extract rows at positions 0, 1, and 2, excluding the upper bound."

".loc is label-based. It selects data using the actual row and column names. For example, df.loc['mango':'sapota'] will extract everything from 'mango' to 'sapota'. Crucially, .loc includes the upper bound because it’s text-based, and it wouldn't make sense to guess what comes 'just before' a text label.

```python

s.iloc[0:2]
```

loc:label based indexing

```python
s.loc[['mango','sapota']]
```

```python
#Broadcasting
df["Salaries"]=df["Salaries"]+5000 #broadcasted the value to match the column with size and shape of the particular column
df["Salaries"]
```

```python
df["Salaries"].unique()
```

```python
df.rename(columns={"Role":"Dept"},inplace=True)
```

```python
#creating new column
df["Promoted Salary"]=df["Salaries"] *5
df
```

```python
#data cleaning to apply ML concepts on data,and perform other calculations smoothly,accurately
df.isnull().sum()
```

```python
#deleting the null value through dropna function,it takes how as parameter where we can assign it with "all"(only deleted the row and column if all the values of that particular rows and columns are NaN) and "any"(deletes even if one row or column has Nan value)
df.dropna(how="all")
```

```python
#filling the nan values(missing values)
#we can get rid of this missing values using two methods ,either deleting the row entirly and second one is filling it with value(using fillna method and forward fill,backward fill)
df["Age"].fillna(df["Age"].mean())
```

```python
df["Salaries"].fillna(df["Salaries"].median())
```

```python
df["Age"].fillna(method='ffill')
```

```python
df["Age"].fillna(method="bfill")
```

```python
#replacing the column value
df["Names"]=df["Names"].replace("Shashi","Shib")
```

```python
#Dealing with duplicate values
df_dup=df[df.duplicated(keep='first')]
df_dup
```

```python
#lamda function
df["Promoted Salary"]=df["Promoted Salary"].apply(lambda x:x/10 if x>65000 else x)
df
```

```python
df["first_last_names"]=["Bhargavi Yaganti","Devi Yaganti","Shib dev","Bhargavi Yaganti"]
df
```

```python
#apply and lambda
def multiply(x):
    return x*2
df["Age"]=df["Age"].apply(multiply)
df
```

```python
#joins and merge
roles={
    "Dept":["ML","DS","Web development","ML"],
    "Location":["New York","San Francisco","Chicago","India"],
    "Manager":["Laura","Steve","Nina","Heena"],
    "Date":[20-3-26,21-4-26,22-5-26,27-0-26]
}
df1=pd.DataFrame(roles)
pd.concat([df,df1],axis=1)
```
